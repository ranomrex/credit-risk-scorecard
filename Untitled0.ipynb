```python
import pandas as pd

df = pd.read_csv("cs-training.csv", index_col=0)
print(df.shape)
print(df.head())
print(df.isnull().sum())
print(df["SeriousDlqin2yrs"].mean())
```

```python
pd.set_option("display.float_format", lambda x: f"{x:.2f}")

# Summary of every column
print(df.describe().T)

# Default rate by age group
df["age_band"] = pd.cut(df["age"], bins=[0, 30, 40, 50, 60, 70, 120])
print(df.groupby("age_band", observed=True)["SeriousDlqin2yrs"].mean())

# Default rate by number of 90-day late payments
print(df.groupby("NumberOfTimes90DaysLate")["SeriousDlqin2yrs"].mean())
```

```python

# 1. Remove impossible age
df = df[df["age"] > 0]

# 2. Handle error codes 96/98: flag them, then reset to 0
late_cols = ["NumberOfTime30-59DaysPastDueNotWorse",
             "NumberOfTime60-89DaysPastDueNotWorse",
             "NumberOfTimes90DaysLate"]
df["late_code_error"] = (df["NumberOfTimes90DaysLate"] >= 96).astype(int)
for c in late_cols:
    df.loc[df[c] >= 96, c] = 0

# 3. Fill missing values, and flag missing income
df["income_missing"] = df["MonthlyIncome"].isnull().astype(int)
df["MonthlyIncome"] = df["MonthlyIncome"].fillna(df["MonthlyIncome"].median())
df["NumberOfDependents"] = df["NumberOfDependents"].fillna(0)

# 4. Cap extreme values at the 99th percentile
for c in ["RevolvingUtilizationOfUnsecuredLines", "DebtRatio", "MonthlyIncome"]:
    df[c] = df[c].clip(upper=df[c].quantile(0.99))

# 5. Remove helper column and check
df = df.drop(columns=["age_band"])
print(df.shape)
print("Missing values left:", df.isnull().sum().sum())
print(df.describe().T[["min", "max"]])
print(df.groupby("income_missing")["SeriousDlqin2yrs"].mean())
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

# 1. Shrink skewed columns
for c in ["DebtRatio", "MonthlyIncome"]:
    df[c] = np.log1p(df[c])

# 2. Separate inputs (X) and answer (y)
X = df.drop(columns=["SeriousDlqin2yrs"])
y = df["SeriousDlqin2yrs"]

# 3. Split: 70% to learn, 30% to test
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y)

# 4. Put all columns on the same scale
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_test_s = scaler.transform(X_test)

# 5. Train logistic regression
model = LogisticRegression(max_iter=1000)
model.fit(X_train_s, y_train)

# 6. Which factors push risk up or down
coef = pd.Series(model.coef_[0], index=X.columns).sort_values()
print(coef)

# 7. Predicted default probability for first 10 test customers
test_prob = model.predict_proba(X_test_s)[:, 1]
print(test_prob[:10].round(3))
```

```python
from sklearn.metrics import roc_auc_score, roc_curve
import matplotlib.pyplot as plt

train_prob = model.predict_proba(X_train_s)[:, 1]
auc_train = roc_auc_score(y_train, train_prob)
auc_test = roc_auc_score(y_test, test_prob)
print("AUC train:", round(auc_train, 3))
print("AUC test:", round(auc_test, 3))
print("Gini test:", round(2 * auc_test - 1, 3))

fpr, tpr, _ = roc_curve(y_test, test_prob)
print("KS test:", round(max(tpr - fpr), 3))

plt.plot(fpr, tpr, label=f"AUC = {auc_test:.3f}")
plt.plot([0, 1], [0, 1], "--", color="grey")
plt.xlabel("Good customers wrongly flagged")
plt.ylabel("Defaulters caught")
plt.title("ROC curve")
plt.legend()
plt.show()
```

```python
results = pd.DataFrame({"prob": test_prob, "default": y_test.values})
results = results.sort_values("prob").reset_index(drop=True)
total_bads = results["default"].sum()

rows = []
for rate in [0.50, 0.60, 0.70, 0.80, 0.85, 0.90, 0.95, 1.00]:
    approved = results.iloc[:int(len(results) * rate)]
    bads = approved["default"].sum()
    goods = len(approved) - bads
    profit = goods * 5000 - bads * 50000
    rows.append({
        "approval_rate": f"{rate:.0%}",
        "cutoff_prob": round(approved["prob"].max(), 3),
        "bad_rate": f"{approved['default'].mean():.2%}",
        "defaulters_rejected": f"{1 - bads / total_bads:.0%}",
        "profit_cr": round(profit / 1e7, 2),
    })

table = pd.DataFrame(rows)
print(table.to_string(index=False))
```
